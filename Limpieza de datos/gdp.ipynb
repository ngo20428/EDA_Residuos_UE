```python
import pandas as pd
import numpy as np
```

```python
# guarda en variables los datasets y su fuente
gdp_cap = pd.read_csv('../Data/Original/sdg_08_10_page_linear.csv.gz', compression='gzip')
```

```python
gdp_cap
```

```python
gdp_cap_clean = gdp_cap.copy()

# Eliminar columnas innecesarias
gdp_cap_clean = gdp_cap_clean.drop(
    columns=['DATAFLOW', 'LAST UPDATE', 'freq', 'unit', 'na_item']
)

# Renombrar columnas
gdp_cap_clean = gdp_cap_clean.rename(columns={
    'TIME_PERIOD': 'year',
    'OBS_VALUE': 'gdp_pc',
    'OBS_FLAG': 'obs_flag',
    'CONF_STATUS': 'conf_status',
})
```

```python
paises_ue = [
    "Austria", "Belgium", "Bulgaria", "Croatia", "Cyprus", 
    "Czechia", "Denmark", "Estonia", "Finland", 
    "France", "Germany", "Greece", "Hungary", "Ireland", 
    "Italy", "Latvia", "Lithuania", "Luxembourg", "Malta", 
    "Netherlands", "Poland", "Portugal", "Romania", "Slovakia", 
    "Slovenia", "Spain", "Sweden", 'European Union - 27 countries (from 2020)'
]

gdp_cap_clean = gdp_cap_clean[gdp_cap_clean['geo'].isin(paises_ue)]
```

```python
PAISES_ES = {'Austria': 'Austria', 'Belgium': 'Bélgica', 'Bulgaria': 'Bulgaria', 'Croatia': 'Croacia', 'Cyprus': 'Chipre', 'Czechia': 'Chequia', 'Denmark': 'Dinamarca', 'Estonia': 'Estonia', 'Finland': 'Finlandia', 'France': 'Francia', 'Germany': 'Alemania', 'Greece': 'Grecia', 'Hungary': 'Hungría', 'Ireland': 'Irlanda', 'Italy': 'Italia', 'Latvia': 'Letonia', 'Lithuania': 'Lituania', 'Luxembourg': 'Luxemburgo', 'Malta': 'Malta', 'Netherlands': 'Países Bajos', 'Poland': 'Polonia', 'Portugal': 'Portugal', 'Romania': 'Rumanía', 'Slovakia': 'Eslovaquia', 'Slovenia': 'Eslovenia', 'Spain': 'España', 'Sweden': 'Suecia', 'European Union - 27 countries (from 2020)': 'Unión Europea (UE-27)'}
```

```python
gdp_cap_clean = gdp_cap_clean.copy()

gdp_cap_clean['pais'] = gdp_cap_clean['geo'].replace(PAISES_ES)
```

```python
gdp_cap_clean
```

```python
gdp_cap_reclean = gdp_cap_clean[["pais", "year", "gdp_pc"]]
```

```python
gdp_cap_reclean.to_csv("../Data/gdp_pc.csv", index=False)
```

### PIB per capita de 2024

```python
gdp_pc_2024 = gdp_cap_reclean[gdp_cap_reclean["year"] == 2024].copy()
```

```python
gdp_pc_2024 = gdp_pc_2024.drop(columns = ["year"])
```

```python
gdp_pc_2024.to_csv("../Data/gdp_pc_2024.csv", index=False)
```
